```python
import numpy as np
import matplotlib.pyplot as plt

# 读取数据
data = np.loadtxt('/global/cfs/cdirs/m3578/chenbo/nanocrystal/CdTe/sq/DOS_line.txt')
energy = data[:, 0]
pdos1 = data[:, 2]  # 片段1: CdTe
pdos2 = data[:, 3]  # 片段2: Cl

# 堆叠画图
plt.figure(figsize=(5, 7))
plt.fill_betweenx(energy, 0, pdos1, color='red', label='CdTe')
plt.fill_betweenx(energy, pdos1, pdos1+pdos2, color='green', label='Cl')

plt.xlabel('Density-of-states')
plt.ylabel('Energy (eV)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 读取数据
data = np.loadtxt('/global/cfs/cdirs/m3578/chenbo/nanocrystal/CdTe/sq/DOS_curve.txt')
energy = data[:, 0]
pdos1 = data[:, 2]  # CdTe
pdos2 = data[:, 3]  # Cl

plt.figure(figsize=(5, 7))

# 画每个能级的横线，叠加
for i in range(len(energy)):
    # CdTe
    plt.hlines(energy[i], 0, pdos1[i], color='red', linewidth=1)
    # Cl，起点是pdos1[i]，终点是pdos1[i]+pdos2[i]
    plt.hlines(energy[i], pdos1[i], pdos1[i]+pdos2[i], color='green', linewidth=1)

plt.xlabel('Density-of-states')
plt.ylabel('Energy (eV)')
plt.legend(['CdTe', 'Cl'])
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 读取数据
data = np.loadtxt('/global/cfs/cdirs/m3578/chenbo/nanocrystal/CdTe/sq/DOS_line.txt')
energy = data[:, 0]
pdos1 = data[:, 2]  # CdTe
pdos2 = data[:, 3]  # Cl

plt.figure(figsize=(4, 6))

for i in range(len(energy)):
    # 跳过 -3.8 到 -2.8 eV 区间
    if -3.8 < energy[i] < -2.8:
        continue
    # CdTe
    plt.hlines(energy[i], 0, pdos1[i], color='red', linewidth=1)
    # Cl
    plt.hlines(energy[i], pdos1[i], pdos1[i]+pdos2[i], color='green', linewidth=1)

plt.xlabel('Density-of-states')
plt.ylabel('Energy (eV)')
plt.xlim(left=0)
plt.legend(['CdTe', 'Cl'], loc='lower right')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 读取数据
data = np.loadtxt('/global/cfs/cdirs/m3578/chenbo/nanocrystal/CdTe/sq/DOS_line.txt')
energy = data[:, 0]
pdos1 = data[:, 2]  # CdTe
pdos2 = data[:, 3]  # Cl

plt.figure(figsize=(2.5, 7))  # 竖直长条

for i in range(len(energy)):
    # 只画 -6 到 -1 eV 区间
    if energy[i] < -6 or energy[i] > -1:
        continue
    # CdTe
    plt.hlines(energy[i], 0, pdos1[i], color='red', linewidth=1)
    # Cl
    plt.hlines(energy[i], pdos1[i], pdos1[i]+pdos2[i], color='green', linewidth=1)

plt.xlabel('Density-of-states')
plt.ylabel('Energy (eV)')
plt.xlim(left=0)
plt.ylim(-6, -1)
plt.legend(['CdTe', 'Cl'], loc='lower right')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 读取数据
data = np.loadtxt('/global/cfs/cdirs/m3578/chenbo/nanocrystal/CdTe/sq/DOS_line.txt')
energy = data[:, 0]
pdos1 = data[:, 2]  # CdTe
pdos2 = data[:, 3]  # Cl

fig, ax = plt.subplots(figsize=(2.5, 7))

for i in range(len(energy)):
    if energy[i] < -6 or energy[i] > -1:
        continue
    ax.hlines(energy[i], 0, pdos1[i], color='red', linewidth=1)
    ax.hlines(energy[i], pdos1[i], pdos1[i]+pdos2[i], color='green', linewidth=1)

# 去掉右、上、下边框
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_position(('data', 0))  # y轴在x=0

# 不显示横坐标数值
ax.set_xticks([])
ax.set_xlabel('')

# 纵坐标范围
ax.set_ylim(-6, -1)
ax.set_xlim(left=0)

# 纵坐标箭头
arrowprops = dict(arrowstyle='->', linewidth=1.5, color='black', shrinkA=0, shrinkB=0)
ax.annotate('', xy=(0, -1), xytext=(0, -6), arrowprops=arrowprops)

# 只保留纵坐标标签
ax.set_ylabel('Energy (eV)', fontsize=12)
ax.yaxis.set_label_coords(-0.15, 0.5)

# 图例
ax.legend(['CdTe', 'Cl'], loc='lower right', frameon=False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 读取数据
data = np.loadtxt('/global/cfs/cdirs/m3578/chenbo/nanocrystal/CdTe/sq/DOS_line.txt')
energy = data[:, 0]
pdos1 = data[:, 2]  # CdTe
pdos2 = data[:, 3]  # Cl

fig, ax = plt.subplots(figsize=(2.5, 7))

# 设定省略区间
omit_min, omit_max = -4, -2

# 画省略区间以外的横线
for i in range(len(energy)):
    if (energy[i] < -6) or (energy[i] > -1):
        continue
    if omit_min < energy[i] < omit_max:
        continue
    ax.hlines(energy[i], 0, pdos1[i], color='red', linewidth=1)
    ax.hlines(energy[i], pdos1[i], pdos1[i]+pdos2[i], color='green', linewidth=1)

# 画省略符号（两条斜线）
x0, x1 = 0, max(pdos1+pdos2)*0.7
y_omit1 = omit_min + 0.15
y_omit2 = omit_min + 0.35
ax.plot([x0, x1], [y_omit1, y_omit2], color='k', linewidth=2)
ax.plot([x0, x1], [y_omit1-0.15, y_omit2-0.15], color='k', linewidth=2)

# 去掉右、上、下边框
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_position(('data', 0))

# 不显示横坐标数值
ax.set_xticks([])
ax.set_xlabel('')

# 纵坐标范围
ax.set_ylim(-6, -1)
ax.set_xlim(left=0)

# 纵坐标箭头
arrowprops = dict(arrowstyle='->', linewidth=1.5, color='black', shrinkA=0, shrinkB=0)
ax.annotate('', xy=(0, -1), xytext=(0, -6), arrowprops=arrowprops)

# 只保留纵坐标标签
ax.set_ylabel('Energy (eV)', fontsize=12)
ax.yaxis.set_label_coords(-0.15, 0.5)

# 图例
ax.legend(['CdTe', 'Cl'], loc='lower right', frameon=False)

plt.tight_layout()
plt.show()
```
