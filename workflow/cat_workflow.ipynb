## Build Quantum Dot

Specify your core structure, ligand(s), and attachment parameters.

### Ligand types

| Type | Examples | Binds to | `dummy_atom` |
|------|----------|----------|--------------|
| **L** | PR₃, RNH₂ | surface cation (e.g. Cd) | cation |
| **X** | Cl⁻, RCOO⁻ | surface cation (e.g. Cd) | cation |
| **Z** | Cd(OOCR)₂ | surface anion (e.g. Te)  | anion  |

### CAT anchor convention

CAT replaces `dummy_atom` on the core with `anchor_atom` from the ligand.
- For **L/X-type** ligands: the binding atom (e.g. P in PPh₃) is not a metal, so a bare PPh₃ would remove the surface Cd. Set `ligand_anchor_atom="P"` and the code auto-prepends a Cd to the ligand in the lone-pair direction.
- For **Z-type** ligands: the ligand already contains the metal anchor (e.g. Cd in Cd(OOCR)₂), no preprocessing needed.

````python
core_file          = "./structures/core/II-VI/CdTe/Cd68Te55Cl26_HLE17_20ang_OPT.xyz"
ligand_files       = "./structures/ligand/PPh3_Cd.mol"  # pre-built 3D mol with Cd anchor
attachment_indices = [45, 49, 56, 39]                # Attachment site indices on core
ligand_type        = "L"                             # "L", "X", or "Z"
ligand_anchor_atom = None                            # Cd already in mol; no auto-add needed
dummy_atom         = "Cd"                            # Core attachment atom
anchor_atom        = "Cd"                            # Ligand anchor atom
alignment          = "surface"   # "surface" (recommended) or "sphere"
subset_f           = None        # fraction of surface sites to fill, e.g. 0.5 for 50%
subset_mode        = "uniform"   # "uniform" (evenly spread), "cluster", or "random"
optimize_ligand    = True
optimize_qd        = False
output_dir         = "./QD_output"

build_quantum_dot(
    core_file          = core_file,
    ligand_files       = ligand_files,
    attachment_indices = attachment_indices,
    ligand_type        = ligand_type,
    ligand_anchor_atom = ligand_anchor_atom,
    dummy_atom         = dummy_atom,
    anchor_atom        = anchor_atom,
    alignment          = alignment,
    subset_f           = subset_f,
    subset_mode        = subset_mode,
    optimize_ligand    = optimize_ligand,
    optimize_qd        = optimize_qd,
    output_dir         = output_dir,
)
````
