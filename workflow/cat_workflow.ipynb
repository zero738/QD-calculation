## Build Quantum Dot

Specify your core structure, ligand(s), and attachment parameters.

**Tips:**
- `attachment_indices`: atom indices on the core where ligands will be attached
- `dummy_atom`: atom type used as the anchor point **on the core** (typically the metal, e.g. `"Cd"`)
- `anchor_atom`: atom type on the **ligand** that binds to the core (e.g. `"Cd"` in PPh3_Cd.mol)
- `optimize_ligand`: optimize ligand geometry before attachment (recommended: `True`)

````python
core_file          = "../model/CdTe_PPh3/core/Cd68Te55Cl26.xyz"  # Core structure file (.xyz)
ligand_files       = "../model/CdTe_PPh3/ligand/PPh3_Cd.mol"     # Ligand file(s) (.mol)
attachment_indices = [45, 49, 56, 39]                            # Atom indices for ligand attachment
dummy_atom         = "Cd"                                        # Anchor atom type on core
anchor_atom        = "Cd"                                        # Anchor atom type on ligand
optimize_ligand    = True                                        # Optimize ligand before attachment
optimize_qd        = False                                       # Optimize full QD after construction
output_dir         = "./QD_output"                               # Output directory

build_quantum_dot(
    core_file          = core_file,
    ligand_files       = ligand_files,
    attachment_indices = attachment_indices,
    dummy_atom         = dummy_atom,
    anchor_atom        = anchor_atom,
    optimize_ligand    = optimize_ligand,
    optimize_qd        = optimize_qd,
    output_dir         = output_dir,
)
````
